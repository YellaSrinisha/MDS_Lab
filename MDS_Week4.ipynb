```python
import pandas as pd

# Series from a list
numbers = [10, 20, 30, 40, 50]
s1 = pd.Series(numbers)
print("Series from List:")
print(s1)

# Series from a dictionary
marks = {"A": 90, "B": 80, "C": 70}
s2 = pd.Series(marks)
print("\nSeries from Dictionary:")
print(s2)

# DataFrame from a list
students = [
    ["Alice", 25],
    ["Bob", 30],
    ["Charlie", 35]
]
df1 = pd.DataFrame(students, columns=["Name", "Age"])
print("\nDataFrame from List:")
print(df1)

# DataFrame from a dictionary
data = {
    "Name": ["David", "Eve", "Frank"],
    "Age": [22, 28, 32]
}
df2 = pd.DataFrame(data)
print("\nDataFrame from Dictionary:")
print(df2)

# DataFrame from Series
col1 = pd.Series([1, 2, 3])
col2 = pd.Series([4, 5, 6])

df3 = pd.DataFrame({
    "Col1": col1,
    "Col2": col2
})

print("\nDataFrame from Series:")
print(df3)
```

```python
import pandas as pd

# Import Employee.csv into a Pandas DataFrame
file_path = '/content/Employee.csv'
df_employee = pd.read_csv(file_path)

print("First 10 records of Employee data.csv:")
print(df_employee.head(10))

print("\nLast 10 records of Employee data.csv:")
print(df_employee.tail(10))
```

```python
print("Shape of the DataFrame:", df_employee.shape)
print("\nIndex of the DataFrame:", df_employee.index)
print("\nColumns of the DataFrame:", df_employee.columns)
```

```python
# Select employees older than 30
print("Employees older than 30:")
print(df_employee[df_employee["Age"] > 30].head())

# Select female employees from Bangalore who joined after 2015
print("\nFemale employees from Bangalore who joined after 2015:")
print(df_employee[
    (df_employee["Gender"] == "Female") &
    (df_employee["City"] == "Bangalore") &
    (df_employee["JoiningYear"] > 2015)
].head())

# Delete rows where LeaveOrNot = 1
new_df = df_employee[df_employee["LeaveOrNot"] == 0]
print("\nAfter removing employees who left:")
print(new_df.head())

# Select one column
print("\nEducation column:")
print(df_employee["Education"].head())

# Select multiple columns
print("\nEducation, City and Age columns:")
print(df_employee[["Education", "City","Age"]].head())

# Delete one column
df1 = df_employee.drop("EverBenched", axis=1)
print("\nAfter dropping EverBenched column:")
print(df1.head())

# Delete multiple columns
df2 = df_employee.drop(["Education", "JoiningYear"], axis=1)
print("\nAfter dropping Education and JoiningYear columns:")
print(df2.head())
```

```python
import pandas as pd

# Sort by Age (ascending)
print("Sort by Age:")
print(df_employee.sort_values("Age").head())

# Sort by Age (descending)
print("\nSort by Age (Descending):")
print(df_employee.sort_values("Age", ascending=False).head())

# Sort by City and Age
print("\nSort by City and Age:")
print(df_employee.sort_values(["City", "Age"]).head())

# Rank by Age
df_employee["Age_Rank"] = df_employee["Age"].rank()
print("\nAge Rank:")
print(df_employee[["Age", "Age_Rank"]].head())

# Rank by Experience (highest experience = rank 1)
df_employee["Exp_Rank"] = df_employee["ExperienceInCurrentDomain"].rank(
    method="dense", ascending=False
)
print("\nExperience Rank:")
print(df_employee[["ExperienceInCurrentDomain", "Exp_Rank"]].head())

# Rank Age within each City
df_employee["City_Age_Rank"] = df_employee.groupby("City")["Age"].rank()
print("\nAge Rank within each City:")
print(df_employee[["City", "Age", "City_Age_Rank"]].head())
```

```python
import pandas as pd

# Describe (includes count, mean, std, min, 25%, 50%, 75%, max)
print("Describe:")
print(df_employee.describe())

# Percentiles
print("\n25th Percentile:")
print(df_employee.quantile(0.25, numeric_only=True))

print("\n50th Percentile (Median):")
print(df_employee.quantile(0.50, numeric_only=True))

print("\n75th Percentile:")
print(df_employee.quantile(0.75, numeric_only=True))

# Mean
print("\nMean:")
print(df_employee.mean(numeric_only=True))

# Median
print("\nMedian:")
print(df_employee.median(numeric_only=True))

# Mode
print("\nMode:")
print(df_employee.mode().head(1))

# Standard Deviation
print("\nStandard Deviation:")
print(df_employee.std(numeric_only=True))

# Variance
print("\nVariance:")
print(df_employee.var(numeric_only=True))

# Correlation
print("\nCorrelation:")
print(df_employee.corr(numeric_only=True))

# Value Counts
print("\nValue Counts (City):")
print(df_employee["City"].value_counts())
```

```python
import pandas as pd

# Categorical columns
cat_cols = df_employee.select_dtypes(include="object").columns

for col in cat_cols:
    print(f"\nColumn: {col}")

    # Number of unique values
    print("Unique Values:", df_employee[col].nunique())

    # Frequency of each value
    print("Value Counts:")
    print(df_employee[col].value_counts())
```

```python
import pandas as pd

# Rename one column
df1 = df_employee.rename(columns={"Age": "EmployeeAge"})
print("After Renaming Age:")
print(df1.head())

# Rename multiple columns
df2 = df_employee.rename(columns={
    "JoiningYear": "YearOfJoining",
    "LeaveOrNot": "LeftCompany"
})
print("\nAfter Renaming Multiple Columns:")
print(df2.head())
```
