```python
import pandas as pd

df = pd.read_csv("/content/Employee.csv")

print("DataFrame head:")
print(df.head())
```

```python
print("Missing values before handling:")
print(df.isnull().sum())

# Check for missing values
if df.isnull().sum().sum() > 0:
    print("\nMissing values detected. Dropping rows with missing values.")

    df_cleaned = df.dropna()

    print("Original shape:", df.shape)
    print("Cleaned shape:", df_cleaned.shape)

    # Update DataFrame
    df = df_cleaned.copy()

    print("\nMissing values after handling:")
    print(df.isnull().sum())

else:
    print("\nNo missing values found in the DataFrame.")
```

```python
print("Detecting outliers using IQR method...")

# Select numerical columns
numerical_cols = df.select_dtypes(include=['number']).columns

# Store original shape
original_shape = df.shape

# Create mask for outliers
outlier_mask = pd.Series(False, index=df.index)

# Find outliers using IQR
for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outlier_mask = outlier_mask | (df[col] < lower) | (df[col] > upper)

# Remove outliers
df_cleaned = df[~outlier_mask].copy()

print("Original shape:", original_shape)
print("Cleaned shape:", df_cleaned.shape)
print("Number of outliers removed:",
      original_shape[0] - df_cleaned.shape[0])

# Update DataFrame
df = df_cleaned.copy()

print("DataFrame updated after outlier removal.")
```
