```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# Load the image
img = Image.open('/content/cat image.jfif')

# Convert image to NumPy array
img_array = np.array(img)

# Crop only the cat's eyes
# Format: image[start_row:end_row, start_column:end_column)
eyes = img_array[135:190, 145:285]

# Display the original image
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")
plt.show()

# Display the cropped eyes
plt.imshow(eyes)
plt.title("Cropped Eyes")
plt.axis("off")
plt.show()
```

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# Load the image
img = Image.open('/content/cat image.jfif')

# Convert image to NumPy array
img_array = np.array(img)

# Perform image transformations
horizontal_flip = np.fliplr(img_array)            # Horizontal Flip
vertical_flip = np.flipud(img_array)              # Vertical Flip
rotate_90 = np.rot90(img_array, 1)                # Rotate 90°
horizontal_of_vertical = np.fliplr(vertical_flip) # Horizontal Flip of Vertical Image
rotate_270 = np.rot90(img_array, 3)               # Rotate 270°

# Create a 3x2 subplot layout
plt.figure(figsize=(12, 15))

# 1.1 Original Image
plt.subplot(3, 2, 1)
plt.imshow(img_array)
plt.title("1.1 Original")
plt.axis("off")
plt.gca().set_aspect("equal")

# 1.2 Horizontal Flip
plt.subplot(3, 2, 2)
plt.imshow(horizontal_flip)
plt.title("1.2 Horizontal Flip")
plt.axis("off")
plt.gca().set_aspect("equal")

# 2.1 Vertical Flip
plt.subplot(3, 2, 3)
plt.imshow(vertical_flip)
plt.title("2.1 Vertical Flip")
plt.axis("off")
plt.gca().set_aspect("equal")

# 2.2 Rotate 90°
plt.subplot(3, 2, 4)
plt.imshow(rotate_90)
plt.title("2.2 Rotate 90°")
plt.axis("off")
plt.gca().set_aspect("equal")

# 3.1 Horizontal Flip of Vertical Image
plt.subplot(3, 2, 5)
plt.imshow(horizontal_of_vertical)
plt.title("3.1 Horizontal Flip of Vertical")
plt.axis("off")
plt.gca().set_aspect("equal")

# 3.2 Rotate 270°
plt.subplot(3, 2, 6)
plt.imshow(rotate_270)
plt.title("3.2 Rotate 270°")
plt.axis("off")
plt.gca().set_aspect("equal")

# Adjust spacing between subplots
plt.tight_layout(pad=2)

# Display all images
plt.show()
```

### Using OpenCV (`cv2`) for Image Operations

Now, let's perform image cropping and flipping using the `cv2` library. First, we need to import it.

```python
import cv2
import matplotlib.pyplot as plt

# Load the image using cv2.imread
# cv2 reads images in BGR format by default
image_path = '/content/cat image.jfif'
img_bgr = cv2.imread(image_path)

# Check if image was loaded successfully
if img_bgr is None:
    print(f"Error: Could not load image from {image_path}")
else:
    # Convert BGR to RGB for correct display with matplotlib
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

    # Display the images
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(img_bgr) # Display BGR image (colors might look off)
    plt.title('Original Image (BGR - cv2 default)')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(img_rgb) # Display RGB image (correct colors)
    plt.title('Original Image (RGB - converted)')
    plt.axis('off')

    plt.show()
```

```python
import numpy as np

data = np.genfromtxt("iris.csv", delimiter=",", skip_header=1,usecols=(0,1,2,3))

print(data)
```

```python
import numpy as np

# Assuming 'data' is already loaded from iris.csv

print("Statistical Operations on Iris Data (first column only):")
print("----------------------------------------------------")

# Select the first column (index 0)
# 'axis=0' refers to operations along rows, but for a single column (1D array),
# statistical functions like mean, median, etc., operate over all elements naturally.
column_data = data[:, 0]

print(f"\nColumn 1 Statistics:")
print(f"  Mean: {np.mean(column_data):.2f}")
print(f"  Median: {np.median(column_data):.2f}")
print(f"  Standard Deviation: {np.std(column_data):.2f}")
print(f"  Minimum: {np.min(column_data):.2f}")
print(f"  Maximum: {np.max(column_data):.2f}")
```

```python
import numpy as np

# Assuming 'data' is already loaded and 'column_data' is defined from the previous cell
# column_data = data[:, 0]

print("Comparison Operations on Iris Data (first column only):")
print("----------------------------------------------------")

# Define a threshold for comparison (e.g., sepal length greater than 5.0)
threshold = 5.0

# Perform comparison operation: check which values are greater than the threshold
greater_than_threshold_mask = column_data > threshold

print(f"\nValues in Column 1 greater than {threshold}:")
print(column_data[greater_than_threshold_mask])

# Count how many values meet the condition
count_greater = np.sum(greater_than_threshold_mask)
print(f"\nNumber of values in Column 1 greater than {threshold}: {count_greater}")

# Another comparison: check which values are less than or equal to a different threshold
threshold_le = 4.8
less_than_or_equal_to_threshold_mask = column_data <= threshold_le
print(f"\nValues in Column 1 less than or equal to {threshold_le}:")
print(column_data[less_than_or_equal_to_threshold_mask])

count_less_equal = np.sum(less_than_or_equal_to_threshold_mask)
print(f"\nNumber of values in Column 1 less than or equal to {threshold_le}: {count_less_equal}")
```
