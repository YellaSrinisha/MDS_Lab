```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
img=Image.open('image1.jpg')
arr=np.array(img)
plt.imshow(arr)
plt.title('array of image')
plt.show()
cropped_image=arr[50:200,50:200]
plt.imshow(cropped_image)
plt.title('cropped image')
plt.show()
horiz_img=arr[:,::-1]
plt.imshow(horiz_img)
plt.title('horizontal image')
plt.show()
vert_img=arr[::-1,:]
plt.imshow(vert_img)
plt.title('vertical image')
plt.show()
```

```python
import pandas as pd

df = pd.read_csv('House Price Prediction Dataset.csv')
df.columns
plt.figure()
plt.scatter(df['Area'],df['Price'],c=df['Bedrooms'],cmap='viridis')
plt.colorbar()
plt.xlabel('Area')
plt.ylabel('Price')
plt.show()
plt.figure()
plt.scatter(df['Area'],df['Price'],c=df['Bedrooms'],cmap='coolwarm')
plt.colorbar()
plt.xlabel('Area')
plt.ylabel('Price')
plt.show()
plt.figure()
plt.scatter(df['Area'],df['Price'],c=df['Bedrooms'],cmap='jet')
plt.colorbar()
plt.xlabel('Area')
plt.ylabel('Price')
plt.show()
plt.figure()
plt.scatter(df['Area'],df['Price'],c=df['Bedrooms'],cmap='plasma')
plt.colorbar()
plt.xlabel('Area')
plt.ylabel('Price')
plt.show()
```

```python
import pandas as pd
import scipy.stats as stats
df = pd.read_csv('House Price Prediction Dataset.csv')
df.columns
plt.figure()
stats.probplot(df['Area'],dist='norm',plot=plt)
plt.xlabel('Area')
plt.ylabel('Price')
plt.show()
plt.figure()
stats.probplot(df['Price'],dist='norm',plot=plt)
plt.show()
plt.figure()
stats.probplot(df['Bedrooms'],dist='norm',plot=plt)
plt.show()
```
